**Question 1:** Sum of List Elements

```python
def sum_lst(lst):
    m=0
    for i in lst:
        m = m+i
    return m

print(f'Sum of list: {sum_lst([1,2,3,4,5,6])}')
```

**Question 4:** Check if all elements in a List are Unique

```python
def check_unique(lst):
    if len(lst) == len(set(lst)):
        return True
    else:
        return False
numbers = [3, 8, 2, 10, 5]
print(check_unique(numbers))
```

**Question 6:** Number of Odd and Even Elements in a list

```python
def count_even_odd(lst):
    even=0
    odd=0
    for i in lst:
        if i%2 ==0:
            even +=1
        else:
            odd +=1
    
    return (even,odd)

numbers = [3, 8, 2, 10, 5]
print(count_even_odd(numbers))
```

**Question 8:** Maximum Difference between 2 consecutive elements in a List

```python
def max_consecutive_difference(lst):
    max = 0
    for i in range(1,len(lst)):
        dif = abs(lst[i] - lst[i-1])
        
        if max < dif:
            max = dif
    
    return max
        
lst = [10, 15, 8, 20, 12]

print(max_consecutive_difference(lst))
```

**Question 10:** You are given a list of integers and an integer k. Write a Python function to rotate the list to the right by k positions without using slicing. A rotation shifts elements from the end of the list to the front.

```python
def rotate_list(lst, k):
    if len(lst) == 0:
        return lst

    k = k%len(lst)

    for _ in range(k):
        last = lst.pop()
        lst.insert(0,last)
    return lst

lst = [10, 15, 8, 20, 12]

print(rotate_list(lst,2))
```

**Question 11:** Design a Python function named merge_lists_to_dictionary to merge two lists into a dictionary where elements from the first list act as keys and elements from the second list act as values.

```python
def merge_lists_to_dictionary(keys, values):

    
    if(len(keys) != len(values)):
        return False
    
    dic = {}
    
    for i in range (0, len(keys)):
        dic[keys[i]] = values[i]
    
    return dic

keys = ["name", "age", "city"]
values = ["Tushar", 22, "Delhi"]

print(merge_lists_to_dictionary(keys, values))
```

**Question 12:** Design a Python function named merge_three_dictionaries to merge exactly three dictionaries into one.

```python
def merge_three_dictionaries(dict1, dict2, dict3):
    
    merge = dict1 | dict2 | dict3
    return merge
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}
dict3 = {"e": 5, "f": 6}
print(merge_three_dictionaries(dict1,dict2,dict3))
```

**Question 13:** Design a Python function named count_word_frequency to count the frequency of words in a sentence and store the counts in a dictionary.

```python
def count_word_frequency(sentence):
    
    words = sentence.split()
    freq = {}
    
    for word in words:
        if word in freq :
            freq[word] += 1
        else:
            freq[word] = 1

    return freq
sentence = "hello world hello"

print(count_word_frequency(sentence))
```

**Question 14:** Design a Python function named is_palindromic_tuple to check if a tuple is palindromic, meaning it reads the same forwards and backwards.

```python
def is_palindromic_tuple(tup):
    # Your code goes here
    if tup == tup[::-1]:
        return True
    else:
        return False

num = (1, 2, 3, 2, 1)
print(is_palindromic_tuple(num))
```

**Question 7:** Design a Python function named merge_dicts_with_overlapping_keys that merges multiple dictionaries into a single dictionary. If a key appears in more than one dictionary, sum up their values.

```python
def merge_dicts_with_overlapping_keys(*dicts):
    result = {}
    
    for data in dicts:
        for key,value in data.items():
            result[key] = result.get(key,0)+value
    
    
    return result
dict1 = {'a': 10, 'b': 20}
dict2 = {'b': 30, 'c': 40}
dict3 = {'a': 5, 'c': 10}

print(merge_dicts_with_overlapping_keys(dict1, dict2, dict3))
```
